```python
RANDOM_SEED = 42
VERBOSE = True
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from minisom_representation import SomRepresentation
```

```python
# offline training -> Batch-based weight update
# 	[	This algorithm changes the weights once per iteration after considering all the data.	] - MiniSom documentation
som_rep_1 = SomRepresentation.with_derived_params(X_scaled, random_seed=RANDOM_SEED, verbose=VERBOSE)
som_rep_1.fit_offline(X_scaled, num_iteration=20);
```

```python
# online training -> Item-based weight update (depends on use_epochs)
# 	[ 	If use_epochs is False, the weights will be updated num_iteration times. Otherwise they will be updated len(data)*num_iteration times.
# 		If random_order is True, samples are picked in random order. Otherwise the samples are picked sequentially.	] - MiniSom documentation
som_rep_2 = SomRepresentation.with_derived_params(X_scaled, random_seed=RANDOM_SEED, verbose=VERBOSE)
som_rep_2.fit_online(X_scaled, num_iteration=20, use_epochs=True, random_order=True);
```

```python
# access extracted and calculated properties

print("SOM:", som_rep_1.som)
print("Shape of lattice:\t\t\t\t", som_rep_1.lattice_shape_)
print("Rows:\t\t\t\t\t\t\t", som_rep_1.rows_)
print("Columns:\t\t\t\t\t\t", som_rep_1.cols_)
print("Quantization Error (QE):\t\t", som_rep_1.quantization_error)
print("Topographic Error (TE):\t\t\t", som_rep_1.topographic_error)

print("\nSOM:", som_rep_2.som)
print("Shape of lattice:\t\t\t\t", som_rep_2.lattice_shape_)
print("Rows:\t\t\t\t\t\t\t", som_rep_2.rows_)
print("Columns:\t\t\t\t\t\t", som_rep_2.cols_)
print("Quantization Error (QE):\t\t", som_rep_2.quantization_error)
print("Topographic Error (TE):\t\t\t", som_rep_2.topographic_error)
```
