```python
RANDOM_SEED = 42
VERBOSE = True
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from minisom_representation import SomRepresentation, calc_recommended_lattice_sides, calc_initial_sigma
```

```python
# retrieve recommended parameters
d1_rec, d2_rec = calc_recommended_lattice_sides(X_scaled)
sigma_rec = calc_initial_sigma(d1_rec, d2_rec)
print("Recommended parameters: ", f"d1={d1_rec}, d2={d2_rec}, sigma={sigma_rec}")
```

```python
# manually define parameters based on recommendations
d1, d2 = d1_rec * 2, d2_rec * 2
sigma = sigma_rec * 2.5
print("Defined parameters: ", f"d1={d1}, d2={d2}, sigma={sigma}")
```

```python
# fit SOM representation
som_rep = SomRepresentation(d1=d1, d2=d2, sigma=sigma, random_seed=RANDOM_SEED, verbose=VERBOSE)
som_rep.fit(X_scaled)
```
