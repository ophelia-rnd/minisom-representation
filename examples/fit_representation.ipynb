```python
RANDOM_SEED = 42
VERBOSE = True
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from minisom_representation import SomRepresentation
```

```python
# explicitly setting `d1`, `d2`, `sigma`
som_rep = SomRepresentation(d1=5, d2=10, sigma=5.0, random_seed=RANDOM_SEED, verbose=VERBOSE)
som_rep.fit(X_scaled)
```

```python
# deriving `d1`, `d2`, `sigma` from data
som_rep = SomRepresentation.with_derived_params(X_scaled, random_seed=RANDOM_SEED, verbose=VERBOSE)
som_rep.fit(X_scaled)
```

```python
# combination of derived and explicitly set parameters
som_rep = SomRepresentation.with_derived_params(
    X_scaled, sigma=1.0, topology="rectangular", learning_rate=0.1, num_iteration=50,
    decay_function="asymptotic_decay", sigma_decay_function="asymptotic_decay",
    neighborhood_function='gaussian', activation_distance='euclidean',
    random_seed=RANDOM_SEED, verbose=True
)
som_rep.fit(X_scaled)
```

```python
# access extracted and calculated properties
print("SOM:", som_rep.som)
print("Number of components:\t\t\t", som_rep.component_size_)
print("Shape of lattice:\t\t\t\t", som_rep.lattice_shape_)
print("Rows:\t\t\t\t\t\t\t", som_rep.rows_)
print("Columns:\t\t\t\t\t\t", som_rep.cols_)
print("Shape of node weights matrix:\t", som_rep.node_weights_.shape)
print("Shape of distance map:\t\t\t", som_rep.distance_map.shape)
print("Shape of activation map:\t\t", som_rep.activation_map.shape)
print("Sum of activation map:\t\t\t", som_rep.activation_map.sum())
print("Quantization Error (QE):\t\t", som_rep.quantization_error)
print("Topographic Error (TE):\t\t\t", som_rep.topographic_error)
```
